```python
import pandas as pd
from pandas import read_csv
import numpy as np
import matplotlib.pyplot as plt
```

### **Asteroid Orbit Classes: Aten**

```python
df = pd.read_csv('sbdb_query_results.csv')
df
```

```python
a = df['a'].values
e = df['e'].values
i = df['i'].values
```

### **Parámetro de Tisserand**

```python
def T(a, e, i):
  return 1/a + 2*np.cos(i) * (a*(1-e**2))**0.5
```

```python
print(T(a, e, i))
print(max(T(a, e, i)))
print(min(T(a, e, i)))
```

```python
plt.scatter(a,e)
```

```python
valores_T = T(a, e, i)
plt.scatter(a, e, c=valores_T, cmap='viridis')
plt.colorbar()
```
