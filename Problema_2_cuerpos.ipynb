**8). Elementos orbitales de una nave espacial.** Una nave espacial se mueve alrededor de la Tierra con una órbita tal que la distancia al perigeo es de 7000 km y la distancia al afelio es de 70000 km. Calcule:

**(a)** La excentricidad de la órbita

**(b)** El semieje mayor en kilómetros

**(c)** El período orbital en horas

**(d)** La energía específica de la órbita en $km^2/s^2$

**(e)** La anomalía verdadera en grados cuando la altitud (sobre la superficie terrestre) es de 1000 km

**(f)** La velocidad en el perigeo y en el apogeo en $km/s$

Para solucionar el punto, primero se definen los valores iniciales de la distancia al afelio $r_a$ y la distancia al perigeo $r_p$, estas distancias están en km. Además se definen la constante $μ$ y $r$ que corresponde al radio terrestre en km.

```python
ra = 7000 # km
rp = 70000 # km
mu = 398600.4418  # km^3/s^2
r = 6378.137  # km

# Conversion de radianes a grados
dec = 180 / np.pi
```

Las fórmulas que se usan para calcular los valores del enunciado son:

Excentricidad de la órbita: $$ e = \frac {r_a - r_p}{r_a + r_p} $$

Semieje mayor: Es el promedio de las distancias al afelio y perigeo. $$ a = \frac {r_a +r_p}{2} $$

Periodo orbital: Se usa la tercera ley de Kepler. $$ T = 2 \pi \sqrt{\frac {a^3}{\mu}} $$

Energía específica: $$ ϵ = \frac{-\mu}{2a} $$

Anomalía verdadera: $$ f = \cos^{-1} \left[ \frac{\frac{a(1 - e^2)}{r + 1000} - 1}{e} \right]$$

Velocidad del perigeo: $$ v_p = \sqrt{\frac{\mu(1+e)}{a(1-e)}} $$

Velocidad del apogeo: $$ v_a = \sqrt{\frac{\mu(1-e)}{a(1+e)}} $$

```python
e = abs((ra - rp) / (ra + rp))

a = (ra + rp) / 2

T = 2 * np.pi * (a**3 / mu)**0.5
T_horas = T / 3600

energia_esp = -mu / (2 * a)

f = np.arccos(((a * (1 - e**2)) / (r + 1000) - 1) / e)

vp = np.sqrt(mu * (1 + e) / (a * (1 - e)))
va = np.sqrt(mu * (1 - e) / (a * (1 + e)))

print('La excentricidad es: {:.2f}'.format(e))
print('El semieje mayor es: {:.2f} km'.format(a))
print("El periodo orbital es: {:.2f} horas".format(T_horas))
print("La energía específica es: {:.2f} km^2/s^2".format(energia_esp))
print("La anomalía verdadera es: {:.2f} grados".format(f * dec))
print("La velocidad en el perigeo es: {:.2f} km/s".format(vp))
print("La velocidad en el apogeo es: {:.2f} km/s".format(va))
```

**9). Anomalía verdadera y distancia radial de la órbita de un satélite.** La órbita de un satélite alrededor de la Tierra tiene un semieje mayor de 4 radios terrestres y una distancia al perigeo de 1.5 radios terrestres. Encuentre la anomalía verdadera y la distancia radial en $t=4$ horas después del paso por el perigeo.

Se toman los valores iniciales, siendo $a$ la distancia de la órbita, de 4 radios terrestres $r$; $r_p$ la distancia del perigeo $q$, de 1.5 radios terrestres $r$ y $t$ las horas después del paso al perigeo.

```python
r_tierra = 6378.137 * 1000 # km
m_tierra = 5.97219e24 # kg
G = 6.67408e-11 # N*m^2/kg^2
mu = G * m_tierra

a = 4 * r_tierra # m
q = 1.5 * r_tierra # m
t = 4 # horas
t_p = 0 # horas
```

Las fórmulas que se usan para calcular los valores del enunciado son:

Excentricidad: $$ e = \frac{1-r_p}{a} $$

Periodo de la órbita: $$ T = 2 \pi \sqrt{\frac {a^3}{\mu}} $$

Anomalía media: $$ M = \frac{2\pi t}{T} $$

Ecuación de Kepler: $$ M = E - e \sin E $$

- Con la ecuación de kepler hallamos la anomalía excentrica $E$ y se resuelve con la librería `scipy` (`from scipy.optimize import fsolve`)

Anomalía verdadera: $$ f = 2\tan ^{-1}\frac{\sqrt{1+e} }{(1-e) \tan(E/2)} $$

Distancia radial: $$ r_{radial} = a (1-r \cos(E)) $$

```python
e = 1 - q / a

T2 = 2 * np.pi * (a**3 / mu)**0.5
T2_ = T2 / 3600

M = (2 * np.pi)/T2_ * (t -t_p)
```

```python
def ec_kepler(E):
    return E - e * np.sin(E) - M

E = fsolve(ec_kepler, M)[0]
```

```python
nu = 2 * np.arctan(np.sqrt((1 + e) / (1 - e)) * np.tan(E / 2))

r_radial = a * (1 - e * np.cos(E))

print("La excentricidad es: {:.2f}".format(e))
print("El periodo orbital es: {:.2f} s".format(T2_))
print("La anomalía media es: {:.4f} rad".format(M))
print("La anomalía excéntrica es: {:.4f} rad".format(E))
print("----------------------------------")
print("La anomalía verdadera es: {:.4f} grados".format(nu * dec))
print("La distancia radial es: {:.2f} m".format(r_radial))
```

**10). Elementos orbitales a partir del vector de estado.** Un satélite en órbita alrededor de la Tierra tiene una posición instantánea $\vec r = 6075.0 \hat i + 3190.0 \hat j + 0 \hat k$ km y una velocidad $\vec v = -1.957 \hat i + 5.818 \hat j + 3.033 \hat k$ km/s. Determine los elementos orbitales asumiendo que el plano fundamental es el plano del ecuador terrestre.

Para resolver el ejercicio primero se establecen los vectores unitarios $i_u$, $j_u$ y $k_u$, que corresponden a los vectores en dirección $x$, $y$ y $z$ respectivamente.

Luego se define un vector posición $r_{vec}$ y un vector velocidad $v_{vec}$ con los valores indicados en el enunciado; y se obtiene su magnitud en las variables $r_{mag}$ y $v_{mag}$

```python
# Definir constantes
iu = np.array([1, 0, 0])
ju = np.array([0, 1, 0])
ku = np.array([0, 0, 1])

r_vec = np.array([6075.0, 3190.0, 0.0])  # km
v_vec = np.array([-1.957, 5.818, 3.033])  # km/s

r_mag = np.linalg.norm(r_vec)
v_mag = np.linalg.norm(v_vec)
```

Ahora, se calculan los elementos orbitales, siguiendo las formulas vistas en clase:

Momento angular específico: $$ \vec h = \vec r \times \vec v $$

Vector de excentricidad: $$ \vec e = \frac {\vec v \times \vec h}{\mu} - \frac{\vec r}{r} $$

Vector del nodo ascendente: $$ \vec n = \hat k \times \vec h $$

- A estos 3 vecores también se sacó su magnitud usando `np.linalg.norm`

Energía obital específica: $$ ϵ = \frac{v^2}{2} - \frac{\mu}{r} $$

Semieje mayor: $$ a = \frac{-\mu}{2 ϵ} $$

Inclinación: $$ i = \cos^{-1} \frac {\vec h_y}{h} $$

Ascensión recta del nodo ascendente: $$ Ω = \cos^{-1} \frac {\vec n_x}{n} $$

Argumento del perigeo: $$ \cos^{-1} \frac {\vec n \cdot \vec e}{n e}  $$

Anomalía verdadera: $$ f = \cos^{-1} \frac{\vec e \cdot \vec r}{e r} $$


```python
h_vec = np.cross(r_vec, v_vec)
h_mag = np.linalg.norm(h_vec)

mu = 398600.4418  # km^3/s^2
e_vec = (np.cross(v_vec, h_vec) / mu) - (r_vec / r_mag)
e_mag = np.linalg.norm(e_vec)

n_vec = np.cross(ku, h_vec)
n_mag = np.linalg.norm(n_vec)

E = (v_mag**2) / 2 - (mu / r_mag)

a = -mu / (2 * E)

i = (np.arccos(h_vec[2] / h_mag))

if n_mag != 0:
    Omega = (np.arccos(n_vec[0] / n_mag))
    if n_vec[1] < 0:
        Omega = 360 - Omega
else:
    Omega = 0

if e_mag != 0:
    omega = (np.arccos(np.dot(n_vec, e_vec) / (n_mag * e_mag)))
    if e_vec[2] < 0:
        omega = 360 - omega
else:
    omega = 0

nu = (np.arccos(np.dot(e_vec, r_vec) / (e_mag * r_mag)))
if np.dot(r_vec, v_vec) < 0:
    nu = 360 - nu

print("Semi-eje mayor (a): {:.2f} km".format(a))
print("Excentricidad (e): {:.4f}".format(e_mag))
print("Inclinación (i): {:.2f}°".format(i * dec))
print("RAAN (Ω): {:.2f}°".format(Omega * dec))
print("Argumento del perigeo (ω): {:.2f}°".format(omega))
print("Anomalía verdadera (ν): {:.2f}°".format(nu * dec))
```

**11). Satélite en órbita.** Un satélite en órbita elíptica alrededor de la Tierra tiene un periodo de 15.7430 horas y una distancia al perigeo de 12756 km (respecto al centro de la Tierra). En t = 10 horas después del paso por el perigeo calcule:

**(a)** La distancia radial

**(b)** La velocidad

**(c)** La anomalía excéntrica

**(d)** La anomalía verdadera

**(e)** La anomalía media

**(f)** La excentricidad

**(g)** La energía específica

**(h)** La velocidad en el apogeo

**(i)** La distancia al apogeo

**(j)** El tiempo que le falta para llegar al apogeo

Se toman los valores iniciales del periodo, $T$, la distancia al perigeo, $r_p$ y $t$ el tiempo después del paso por el perigeo.

```python
T = 15.7430 * 3600  # s
rp = 12756  # km
t = 10 * 3600  # s
```

Para este punto, se utilizan las fórmulas anteriores para calcular el semieje mayor $a$, la excentricidad $e$, la anomalía media $M$, con la anomalía media se halla $E$, que es la anomalía excentrica, usando `fsolve`.

Luego, con la excentricidad y la anomlía excéntrica se halla la anomalía verdadera $f$

Ahora, se usan las siguientes formulas para hallar los demás valores:

Distancia radial: $$ r =  \frac{a(1-e^2)}{1+e \cos f} $$

Velocidad radial: $$ v = \sqrt{\mu \left( \frac{2}{r} - \frac{1}{a} \right)}
 $$

Para calcular el tiempo hasta el apogeo se calcula la anomalía media en el apogeo que corresponde a $\pi$ y se calcula:

$$ ΔM = M_{apogeo} - M $$

$$ Δt = \frac{ΔM}{n}$$

```python
a = (mu * (T**2) / (4 * np.pi**2))**(1/3)

e = 1 - rp / a

n = 2 * np.pi / T
M = n * t

def kepler_eq(E):
    return E - e * np.sin(E) - M

E = fsolve(kepler_eq, M)[0]

f_ = 2 * np.arctan(np.sqrt((1 + e) / (1 - e)) * np.tan(E / 2))

r = a * (1 - e**2) / (1 + e * np.cos(nu))

v = np.sqrt(mu * (2 / r - 1 / a))

energia_especifica = -mu / (2 * a)

ra = a * (1 + e)
va = np.sqrt(mu * (2 / ra - 1 / a))

M_apogeo = np.pi
delta_M = M_apogeo - M
delta_t = abs(delta_M / n)

print("Excentricidad: {:.2f}".format(e))
print("Anomalía Media: {:.2f} grados".format(M * dec))
print("Anomalía Excéntrica: {:.2f} grados".format(E * dec))
print("Anomalía Verdadera: {:.2f} grados".format(f_ * dec))
print("Distancia radial: {:.2f} km".format(r))
print("Velocidad radial: {:.2f} km/s".format(v))
print("Energía Específica: {:.2f} km²/s²".format(energia_especifica))
print("Distancia al Apogeo: {:.2f} km".format(ra))
print("Velocidad en el Apogeo: {:.2f} km/s".format(va))
print("Tiempo hasta el Apogeo: {:.2f} horas".format(delta_t / 3600))
```
