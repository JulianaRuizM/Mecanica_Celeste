```python
import numpy as np
import pandas as pd
```

Se obtienen los vectores de estado los vectores de estado (vector posición $\vec{r}$ y el vector velocidad $\vec{v}$) para los cuerpos del Sol, Júpiter y Saturno para la fecha de 2024-10-31, a través de Horizons System - NASA (https://ssd.jpl.nasa.gov/horizons/app.html#/)

```python
# Vectores de estado del Sol, Júpiter, Saturno

pos_sol = np.array([-9.223125953892386E+05, -6.971582947371479E+05, 2.789422119026093E+04]) * 1000 # en m
pos_jup = np.array([2.256086993617021E+08, 7.219617753824757E+08, -8.042224237805039E+06]) * 1000 # en m
pos_sat = np.array([1.406956569228385E+09, -3.152916834129006E+08, -5.053584979970349E+07]) * 1000 # en m

vel_sol = np.array([1.183203802762799E-02, -7.576359162400811E-03, -1.817523289842603E-04]) * 1000 # en m/s
vel_jup = np.array([-1.261884072114931E+01, 4.519624847356097E+00, 2.635982390958225E-01]) * 1000 # en m/s
vel_sat = np.array([1.575053643300134E+00, 9.405962361845964E+00, -2.262721192052810E-01]) * 1000 # en m/s
```

```python
# Masas de cada planeta

m_sol = 1.98840987e+24 # en kg
m_jup = 1.89851767e+21 # en Kg
m_sat = 5.68457894e+20 # en Kg
```

Ahora se calcula el momento angular de cada cuerpo, el cual se define así:

$$\vec{L} = ∑_i m_i (\vec{r}_i × \vec{v}_i)$$

donde $m$ son las masas de cada cuerpo, $\vec{r}$ son los vectores posición y $\vec{v}$ son los vectores velocidad.

```python
''' Momentum lineal de los planetas '''

# Momento angular
L_sol = np.cross(pos_sol, vel_sol) * m_sol
L_jup = np.cross(pos_jup, vel_jup) * m_jup
L_sat = np.cross(pos_sat, vel_sat) * m_sat

print("El momento angular del sol es", L_sol, "kg m^2 / s")
print("El momento angular de Júpiter es", L_jup, "kg m^2 / s")
print("El momento angular de Saturno es", L_sat, "kg m^2 / s")
```

```python
# Normalizar momento angular (magnitud escalar)
Lsol_norm = np.linalg.norm(L_sol)
Ljup_norm = np.linalg.norm(L_jup)
Lsat_norm = np.linalg.norm(L_sat)

# Momento angular total
L_total = L_sol + L_jup + L_sat

L_total_norm = np.linalg.norm(L_total)

print("El momento angular total del sistema es", L_total, "kg m^2 / s")
print("El momento angular total normalizado es", L_total_norm, "kg m^2 / s")
print("\n")
print("La norma del vector momento angular para el sol es", Lsol_norm, "kg m^2 / s")
print("La norma del vector momento angular para Júpiter es", Ljup_norm, "kg m^2 / s")
print("La norma del vector momento angular para Saturno es", Lsat_norm, "kg m^2 / s")
```

**Conclusión:** El momento angular total del sistema corresponde casi por completo a la suma de los momentos angulares individuales del Sol, Júpiter y Saturno, dado que son los que aportan más contribución al cálculo por su masa. Esto indica que el comportamiento del sistema es el resultado de las contribuciones conjuntas de los momentos angulares de estos tres cuerpos principales.

Júpiter, con el mayor momento angular, domina la dinámica del sistema, seguido por Saturno. El Sol, en comparación, tiene un impacto menor debido a su baja velocidad de rotación y a que los momentos angulares de los planetas son mucho más grandes debido a sus distancias al Sol. Aunque Saturno está más alejado, su contribución al momento angular es menor que la de Júpiter, ya que su velocidad es inferior.

Por otro lado, el Sol actúa como un contrapeso a los planetas, dado que está relativamente cerca del baricentro del sistema solar.

Ahora, para encontrar la **inclinación del plano invariable de Laplace del sistema Sol-Júpiter-Saturno con respecto al plano de la eclíptica**, creamos un vector normal a la eclíptica, la inclinación esta dada por el ángulo formado entre el momento angular total y el eje z del vector normal, de la siguiente manera:

NOTA: El ángulo entre dos vectores se calcula así

 $$ \cos \theta = \frac{\vec{V_{1}} \cdot \vec{V_{2}}}{\|\vec{V_{1}}\| \|\vec{V_{2}}\|} $$

 donde $\vec{V_1}$ es el momento angular total del sistema y $\vec{V_2}$ es el vector normal a la eclíptica, $ (0,0,1) $.


```python
ecl = np.array([0, 0, 1])
ecl_norm = np.linalg.norm(ecl)

# Se calcula el ángulo entre el vector normal de la eclíptica y L
i = np.arccos(np.dot(L_total, ecl) / (L_total_norm * ecl_norm))
i = np.degrees(i)

print("La inclinación del plano invariable de Laplace del sistema con respecto al plano de la eclíptica es", i, "°")
```

**Resultado:** La inclinación del plano invariable de Laplace del sistema con respecto al plano de la eclíptica es 1.635 °

Para **comparar la ubicación del plano invariable con la ubicación de los planos de las órbitas de Júpiter y Saturno** si se sabe que la inclinación de sus órbitas (de nuevo medidas respecto a la eclíptica) son respectivamente 1◦18′ y 2◦29′; se calcula la diferencia entre el angulo del plano invariable y el plano orbital de cada planeta.

```python
# Inclinación de Júpiter
jup_inc_deg = 1 + 18/60  # 1 grado + 18 minutos / 60 minutos por grado
print("Inclinación de Júpiter en grados decimales:", jup_inc_deg, "°")

# Inclinación de Saturno
sat_inc_deg = 2 + 29/60  # 2 grados + 29 minutos / 60 minutos por grado
print("Inclinación de Saturno en grados decimales:", sat_inc_deg, "°")
```

```python
# Diferencia con Júpiter
dif_jup = abs(i - jup_inc_deg)
print("Diferencia entre el plano invariable y la órbita de Júpiter:", dif_jup, "°")

# Diferencia con Saturno
dif_sat = abs(i - sat_inc_deg)
print("Diferencia entre el plano invariable y la órbita de Saturno:", dif_sat, "°")
```

**Conclusión:** El plano invarible tiene menos inclinación con respecto a la órbita de Júpiter que con la órbita de Saturno, esto se debe a que el cuerpo más masivo es el que indica la dirección del plano invariante en un sistema, en este caso del sistema solar (ya que usamos los cuerpos más pesados y las masas de los demás planetas se podrían despreciar).

Se crea el diccionario *planet_datos* para almacenar los datos de los planetas y se crea otro diccionario *mu_planeta* para almacenar los μ de cada planeta, siendo $μ = GM$, luego se consulta con `pc.consulta_horizonts` los vectores de estado (vector posición $\vec{r}$ y el vector velocidad $\vec{v}$) de cada uno para almacenarlos en el diccionario *planet_datos* para poder consultar cada vector correspondiente al id de su planeta; y por último con `SPICE` se obtienen los μ y se almacenan en el diccionario *mu_planetas*.

```python
planet_datos = {}
mu_planeta = {}

for id in ['Sun',
           'Jupiter Barycenter',
           'Saturn Barycenter']:

  tabla, df, datos = pc.consulta_horizons(
    id=id,
    location='@0',
    epochs='2024-10-31 00:00:00',
    datos='vectors'
  )
  mu_cuerpo = spy.bodvrd(id,'GM',1)[1][0]

  mu_planeta[id] = mu_cuerpo * 1000 # cambio de km^3 / s^2 a m^3 / s^2
  planet_datos[id] = datos
```

Se cambian los diccionarios que contienen los vectores de estado a arrays de `numpy`, luego se almacenan las posiciones de cada planeta en la variable Rs y las velocidades de cada planeta en Vs

```python
Rs = np.array([planet_datos[id][0:3] for id in planet_datos])
Vs = np.array([planet_datos[id][3:6] for id in planet_datos])
```

Se hace una lista por comprensión para almacenar las masas de cada planeta, lo que corresponde a dividir cada μ por la constante gravitacional $G$

```python
masas_planetas = np.array([mu_planeta[id] for id in planet_datos])
masas_planetas = masas_planetas / pc.constantes.G
masas_planetas
```

Se almacena en la variable *L_planetas* el momento angular de cada cuerpo, el cual se define así:

$$\vec{L} = ∑_i m_i (\vec{r}_i × \vec{v}_i)$$

donde $m$ son las masas de cada cuerpo, $\vec{r}$ son los vectores posición y $\vec{v}$ son los vectores velocidad.

```python
L_planetas = np.cross(Rs, Vs) * masas_planetas[:,None]

print("El momento angular para el sol es", L_planetas[0], "kg m^2 / s")
print("El momento angular para Júpiter es", L_planetas[1], "kg m^2 / s")
print("El momento angular para Saturno es", L_planetas[2], "kg m^2 / s")
```

Se almacena el momento angular total del sistema solar en la variable *L_total* sumando componente a componente del momento angular de cada cuerpo.

```python
L_total2 = [sum(L_planetas[:, i]) for i in range(3)]
L_total2 = np.array(L_total2)
print("Momentum angular total del Sistema Sol-Júpiter-Saturno:", L_total2)
```

Ahora, se calcula la norma del momento angular de cada cuerpo y se almacena en *l_norma_planeta*.

```python
l_norma_planeta = []

for planeta in L_planetas:
  norma = np.linalg.norm(planeta)

  l_norma_planeta.append(norma)
```

```python
print("El momento angular normalizado del sol es", l_norma_planeta[0], "kg m^2 / s")
print("El momento angular normalizado de Júpiter es", l_norma_planeta[1], "kg m^2 / s")
print("El momento angular normalizado de Saturno es", l_norma_planeta[2], "kg m^2 / s")
```
