El Hamiltoniano del problema relativo de los dos cuerpos en coordenadas esféricas es:

$$

$$

## Solución

### **Demostración 1.**

Para esta demostración, se parte de la ecuación de Hamilton-Jacobi en coordenadas esféricas, para el problema de dos cuerpos, la cual es:

\begin{equation}
    \frac{1}{2m_r} \left[ \left(\frac{\partial W}{\partial r} \right)^2 + \frac{1}{r^2} \left(\frac{\partial W}{\partial \theta} \right)^2 \right] - \frac{m_r \mu}{r} = E
\end{equation}

Donde:
- $W(r, \theta)$ es la función
- $m_r$ es la masa reducida del sistema
- $\mu = GM$ es el parámetro gravitacional
- $E$ es la energía total del sistema

Luego, usaremos separación de variables. De esta manera, se asume que la función se puede escribir como:

\begin{equation}
    W(r, \theta) = W_r(r) + W_\theta(\theta)
\end{equation}

Sustituyendo en la ecuación de Hamilton-Jacobi, se tiene:

\begin{aligned}
   &\frac{1}{2m_r} \left[ \frac{\partial}{\partial r} (W_r + W_\theta) \right]^2
   + \frac{1}{r^2} \left[ \frac{\partial}{\partial \theta} (W_r + W_\theta) \right]^2
   - \frac{m_r \mu}{r} = E
\end{aligned}

\begin{aligned}
   &\frac{1}{2m_r} \left( \frac{\partial W_r}{\partial r} + \frac{\partial W_\theta}{\partial r} \right)^2
   + \frac{1}{r^2} \left( \frac{\partial W_r}{\partial \theta} + \frac{\partial W_\theta}{\partial \theta} \right)^2
   - \frac{m_r \mu}{r} = E
\end{aligned}

Los términos $\frac{\partial}{\partial r}(w_θ)$ y $\frac{\partial}{\partial θ}(w_r)$ son cero, ya que cada función es independiente de la variable con respecto a la cual se está derivando. Esto es consecuencia directa de la separación de variables en la ecuación de Hamilton-Jacobi.

Así se obtiene:

\begin{equation}
    \frac{1}{2m_r} \left[ \left(\frac{d W_r}{d r} \right)^2 + \frac{1}{r^2} \left(\frac{d W_\theta}{d \theta} \right)^2 \right] - \frac{m_r \mu}{r} = E
\end{equation}

Se multiplica por $2m_r$ toda la ecuación para obtener:

\begin{equation}
    \left(\frac{d W_r}{d r} \right)^2 + \frac{1}{r^2} \left(\frac{d W_\theta}{d \theta} \right)^2 = 2m_r \left( E + \frac{m_r \mu}{r} \right)
\end{equation}

Ahora, se realiza la separación de ecuaciones para $r$ y $\theta$; se usa una constante de separación $P_\theta^2$:

\begin{equation}
    \left(\frac{d W_r}{d r} \right)^2 + \frac{1}{r^2} \left(\frac{d W_\theta}{d \theta} \right)^2 = 2m_r \left( E + \frac{m_r \mu}{r} \right) = P_\theta^2
\end{equation}

Esta ecuación se reescribe como dos ecuaciones separadas:

\begin{equation}
    \left(\frac{d W_r}{d r} \right)^2 = 2m_r \left( E + \frac{m_r \mu}{r} \right) - \frac{1}{r^2} \left(\frac{d W_\theta}{d \theta} \right)^2 = P_\theta^2
\end{equation}

Así, imponemos que la parte angular sea igual a la constante de separación:

\begin{equation}
    \left(\frac{d W_\theta}{d \theta} \right)^2 = P_\theta^2 \quad \Rightarrow \quad \frac{d W_\theta}{d \theta} = P_\theta.
\end{equation}

Y la ecuación radial queda expresada como:

\begin{equation}
    \left(\frac{d W_r}{d r} \right)^2 = 2m_r \left( E + \frac{m_r \mu}{r} \right) - \frac{P_\theta^2}{r^2}.
\end{equation}

Lo siguiente es integrar la ecuación angular así:

\begin{equation}
    \int \frac{d W_\theta}{d \theta} = \int P_\theta → W_\theta = P_\theta \theta
\end{equation}


Y luego, se intrega la ecuación radial

\begin{equation}
   \int \frac{d W_r}{d r} = \int \sqrt{ 2m_r \left( E + \frac{m_r \mu}{r} \right) - \frac{P_\theta^2}{r^2} }.
\end{equation}

Así tenemos:

\begin{equation}
    W_r = \int dr \sqrt{ 2m_r \left( E + \frac{m_r \mu}{r} \right) - \frac{P_\theta^2}{r^2} }.
\end{equation}

Por último, del enunciado tenemos la interpretación de $P_r$ y $P_\theta$

- $P_\theta \equiv m_r h$ representa el momentum angular total y es constante.
- $P_r \equiv E$, la energía total del sistema, también es constante.

Esto confirma que $P_r$ y $P_\theta$ actúan como los nuevos momentos generalizados en la formulación canónica del problema.


$$
\boxed{
\begin{aligned}
    & W_\theta = P_\theta \theta = m_r h \\
    & W_r = \int d r \sqrt{2 m_r\left(P_r+\frac{m_r \mu}{r}\right)-\frac{P_\theta^2}{r^2}}
\end{aligned}
}
$$

### **Demostración 2.**

Las coordenadas canónicas generalizadas $Q_r$ y $Q_θ$ están dadas por:

\begin{equation}
  Q_r = \frac{\partial S}{\partial P_r}, \quad Q_θ = \frac{\partial S}{\partial P_θ}
\end{equation}

donde $S$ es la función generatriz de Hamilton-Jacobi, que en este caso se expresa en términos de la función característica $W$:

$$ S = W = W_r + W_\theta $$

Y de la separación de variables en la ecuación de Hamilton-Jacobi, se sabe que:

\begin{aligned}
    & W_\theta = P_\theta \theta = m_r h \\
    & W_r = \int d r \sqrt{2 m_r\left(P_r+\frac{m_r \mu}{r}\right)-\frac{P_\theta^2}{r^2}}
\end{aligned}

Ahora, se calcula $Q_r$:

\begin{equation}
  Q_r = \frac{\partial S}{\partial P_r} = \frac{\partial W}{\partial P_r} = \frac{\partial W_r}{\partial P_r} + \frac{\partial W_θ}{\partial P_r}
\end{equation}

Sin embargo, $W_\theta$ no depende de $P_r$, pues solo involucra $P_\theta$ y $\theta$, esto significa que

$$
Q_r = \frac{\partial S}{\partial P_r} = \frac{\partial W_r}{\partial P_r}
$$

Por otro lado, tenemos por el teorema de la conservación de la energía que se cumple que:

$$ \frac{\partial S}{∂ t} = -E ⇒ \frac{\partial S}{\partial E} = t$$

Esta relación proviene de la teoría de Hamilton-Jacobi, donde el tiempo puede surgir como una derivada parcial de la función generatriz $S$ respecto a la energía $E$, porque en muchos sistemas se usa una separación de variables en la forma:

$$ S = W - Et $$

Luego, si usamos la convención de que $Q_r$ es derivada de S respecto a $P_r$, pero $S$ tiene una dependencia explícita en $t$, entonces se puede escribir:

$$ Q_r = \frac{\partial S}{\partial P_r} = \frac{\partial W_r}{\partial P_r} - \frac{\partial (Et)}{\partial P_r} $$

Y dado que $E$ no depende de $P_r$, suderivada parcial se reduce simplemente a $-t$, así tenemos:

$$ Q_r = \frac{\partial W_r}{\partial P_r} - t$$

Ahora, se deriva la expresión de $W_r$ con respecto a $P_r$:

$$
\frac{\partial W_r}{\partial P_r} = \int \frac{\partial}{\partial P_r} \left( \sqrt{ 2m_r \left( P_r + \frac{m_r \mu}{r} \right) - \frac{P_\theta^2}{r^2} } \right) dr
$$

La derivada dentro de la integral se resuelve como:

Primero, se aplica la regla de la cadena, se define la función dentro de la raíz:  

$$
g(P_r) = 2m_r \left( P_r + \frac{m_r \mu}{r} \right) - \frac{P_\theta^2}{r^2}
$$

Entonces, la derivada de $f(P_r)$ (que corresponde a toda la función a integrar) con respecto a $P_r$ se calcula usando la regla de la cadena:  

$$
\frac{\partial}{\partial P_r} f(P_r) = \frac{1}{2} \cdot \frac{1}{\sqrt{g(P_r)}} \cdot \frac{\partial g}{\partial P_r}
$$

Luego, se calcula la derivada de $g(P_r)$  

$$
\frac{\partial g}{\partial P_r} = \frac{\partial}{\partial P_r} \left( 2m_r \left( P_r + \frac{m_r \mu}{r} \right) - \frac{P_\theta^2}{r^2} \right)
$$

Como $P_\theta$, $m_r$ y $r$ son constantes con respecto a $P_r$, la derivada es:  

$$
\frac{\partial g}{\partial P_r} = 2m_r
$$

Así, se sustituye en la regla de la cadena  

$$
\frac{\partial}{\partial P_r} \left( \sqrt{ 2m_r \left( P_r + \frac{m_r \mu}{r} \right) - \frac{P_\theta^2}{r^2} } \right) = \frac{1}{2} \cdot \frac{1}{\sqrt{g(P_r)}} \cdot 2m_r
$$

Simplificando se tiene:  

$$
\frac{\partial}{\partial P_r} \left( \sqrt{ 2m_r \left( P_r + \frac{m_r \mu}{r} \right) - \frac{P_\theta^2}{r^2} } \right) = \frac{m_r}{\sqrt{ 2m_r \left( P_r + \frac{m_r \mu}{r} \right) - \frac{P_\theta^2}{r^2} }}
$$

Por lo tanto, integrando con respecto a $dr$, obtenemos:

$$
\frac{\partial}{\partial P_r} \left( \sqrt{ 2m_r \left( P_r + \frac{m_r \mu}{r} \right) - \frac{P_\theta^2}{r^2} } \right) = \frac{m_r dr}{\sqrt{ 2m_r \left( P_r + \frac{m_r \mu}{r} \right) - \frac{P_\theta^2}{r^2} }}
$$

Por lo tanto, reemplazando los terminos en $Q_r$, se tiene:

$$
Q_r = \int \frac{m_r \, dr}{\sqrt{ 2m_r \left( P_r + \frac{m_r \mu}{r} \right) - \frac{P_\theta^2}{r^2} }} - t
$$

Por otro lado, se calcula $Q_\theta$:

Usamos la definición:

$$
Q_\theta = \frac{\partial W}{\partial P_\theta} = \frac{\partial W_\theta}{\partial P_\theta} + \frac{\partial W_r}{\partial P_\theta}
$$

Para $W_\theta$, como:

$$
W_\theta = P_\theta \theta
$$

su derivada con respecto a $P_\theta$ es simplemente:

$$
\frac{\partial W_\theta}{\partial P_\theta} = \theta
$$

Ahora, derivamos $W_r$ con respecto a $P_\theta$:

$$
\frac{\partial W_r}{\partial P_\theta} = \int \frac{\partial}{\partial P_\theta} \left( \sqrt{ 2m_r \left( P_r + \frac{m_r \mu}{r} \right) - \frac{P_\theta^2}{r^2} } \right) dr
$$

Aplicando la regla de la derivada:

$$
\frac{\partial}{\partial P_\theta} \left( \sqrt{ 2m_r \left( P_r + \frac{m_r \mu}{r} \right) - \frac{P_\theta^2}{r^2} } \right) = \frac{- P_\theta / r^2}{\sqrt{ 2m_r \left( P_r + \frac{m_r \mu}{r} \right) - \frac{P_\theta^2}{r^2} }}
$$

Por lo tanto, integrando con respecto a $dr$, obtenemos:

$$
\frac{\partial W_r}{\partial P_\theta} = - \int \frac{P_\theta \, dr}{r^2 \sqrt{ 2m_r \left( P_r + \frac{m_r \mu}{r} \right) - \frac{P_\theta^2}{r^2} }}
$$

Finalmente, sumando ambas contribuciones,

$$
Q_\theta = \theta - \int \frac{P_\theta \, dr}{r^2 \sqrt{ 2m_r \left( P_r + \frac{m_r \mu}{r} \right) - \frac{P_\theta^2}{r^2} }}
$$

De esta manera, concluimos que:


$$
\boxed{
\begin{aligned}
    Q_r &= \int \frac{m_r \, dr}{\sqrt{ 2m_r \left( P_r + \frac{m_r \mu}{r} \right) - \frac{P_\theta^2}{r^2} }} - t, \\
    Q_\theta &= \theta - \int \frac{P_\theta \, dr}{r^2 \sqrt{ 2m_r \left( P_r + \frac{m_r \mu}{r} \right) - \frac{P_\theta^2}{r^2} }}
\end{aligned}
}
$$





### **Demostración 3.**

Para esta demsotración se tiene en cuenta que la solución al problema de los dos cuerpos en el caso de una órbita cerrada (elípitica) es $r = a(1-e\cos E)$.

Además, $E$ (la anomalía excéntrica) cumple la ecuación de Kepler:

$$E-e\sin E = n (t-t_p)$$

o alternativamente que

$$r = p/(1+e\cos f)$$

---

**$→$ Momento generalizado $P_r$**

Por definición, se tiene que $P_r = E$, donde $E$ es la energía total del sistema.

Además, se tiene por definición para una órbita elíptica que:

$$ E = \frac{-m_r \mu}{2a} $$

De este modo, obtenemos

$$
\boxed{
\begin{aligned}
  P_r = \frac{-m_r \mu}{2a}
\end{aligned}
}
$$

**$→$ Momento generalizado $P_θ$**

Se tiene la definición del semilatus rectum, la cual es:

$$ p = \frac{h^2}{\mu} → h^2 = p\mu $$

Además, para una órbita elíptica, el semilatus rectum también es:

$$ p = a(1-e^2) $$

Reemplazando:

$$ h = \sqrt{\mu a(1-e^2)} $$

Con lo demostrado anteriormente, se sabe que $P_\theta = m_r h$, entonces

$$
\boxed{
\begin{aligned}
  P_\theta = m_r \sqrt{\mu a(1-e^2)}
\end{aligned}
}
$$

**$→$ Variable generalizada $Q_r$**

A partir del resultado obtenido en el punto 2, se tiene que:

\begin{equation}
Q_r=  \int \frac{m_r \, dr}{\sqrt{2 m_r \left(P_r + \frac{m_r \mu}{r}\right) - \frac{P_\theta^2}{r^2}}} - t
\end{equation}

Se despeja $t$:

\begin{equation}
t = \int \frac{m_r \, dr}{\sqrt{2 m_r \left(P_r + \frac{m_r \mu}{r}\right) - \frac{P_\theta^2}{r^2}}} - Q_r
\end{equation}

Luego, se sustituye $P_r$ por su expresión en términos de la energía orbital:

\begin{equation}
t = \int \frac{m_r \, dr}{\sqrt{2 m_r \left(\frac{-m_r\mu}{2a} + \frac{m_r \mu}{r}\right) - \frac{m_r^2h^2}{r^2}}} - Q_r
\end{equation}

De esta manera, se puede simplificar el argumento de la raíz así:

\begin{equation}
t = \int \frac{m_r \, dr}{\sqrt{m_r^2\left(2\left(\frac{\mu}{2a} + \frac{\mu}{r}\right) - \frac{h^2}{r^2} \right)}} - Q_r
\end{equation}

Se utilizan las siguientes identidades orbitales:

\begin{align*}
h^2 &= a\mu(1-e^2), & r &= a(1 - e \cos E)
\end{align*}

Se reemplazan en la expresión del argumento de la raíz, así:

\begin{equation}
m_r^2 \left(\frac{-\mu}{a} + \frac{2 \mu}{r} - \frac{a\mu(1-e^2)}{r^2} \right) = \frac{m_r^2\mu}{a} \left(-1 + \frac{2 }{(1-e\cos E)} - \frac{1-e^2}{(1-e\cos E)^2} \right)
\end{equation}

Se multiplica por $\frac{(1-e\cos E)^2}{(1-e\cos E)^2}$ se tiene:

\begin{equation}
\frac{m_r^2\mu}{a(1-e\cos E)^2} \left(-(1-e\cos E)^2 + 1 -2e\cos E -e^2\right)
\end{equation}

Se factorizan términos:

\begin{equation}
 -\frac{m_r^2\mu}{a(1-e\cos E)^2}e^2(\cos^2E + 1) = \frac{m_r^2\mu e^2\sin^2E}{a(1-e\cos E)^2}
\end{equation}

Y ahora, se regresa a la ecuación del tiempo:

\begin{equation}
t = \int \frac{m_r \, dr}{\sqrt{\frac{m_r^2\mu e^2\sin^2E}{a(1-e\cos E)^2}}} - Q_r
\end{equation}

Luego se utiliza la relación $r = a(1 - e \cos E) → dr = a e \sin E \, dE$, y se sustituye $dr$:

\begin{equation}
t = \int \frac{a e \sin E \, dE}{\frac{e \sin E}{1-e\cos E}\sqrt{\frac{\mu}{a}}} - Q_r
\end{equation}

Se reorganizan los términos:

\begin{equation}
t = \sqrt{\frac{a^3}{\mu}} \int (1-e\cos E) dE - Q_r
\end{equation}

Se resuelve la integral:

\begin{equation}
t = \sqrt{\frac{a^3}{\mu}} (E - e \sin E) - Q_r
\end{equation}

Por definición, se tiene que $n = \sqrt{\frac{\mu}{a^3}}$, de esta manera:

\begin{equation}
t = \frac{1}{n} (E - e \sin E) - Q_r
\end{equation}

Luego, se despeja en términos de la ecuación de Kepler:

\begin{equation}
E - e \sin E = n (t + Q_r)
\end{equation}

Por último, se compara con la ecuación de Kepler estándar:

\begin{equation}
n(t+Q_r) = n(t-t_p)
\end{equation}

Así, se puede concluir que:

$$
\boxed{
Q_r = -t_p
}
$$


**$→$ Variable generalizada $Q_θ$**

Se tiene que la parte angular se expresa como (demostrado anteriormente):

$$ W_\theta(\theta) = P_\theta \theta $$

Luego, se deriva con respecto a $P_\theta$, asi se tiene:

$$ Q_\theta = \frac{\partial}{\partial P_\theta} (P_\theta \theta) = \theta $$

Por definición, se tiene que el argumento del periapsis es:

$$ w = \theta - f $$

En este contexto, se observa que el resultado obtenido para $Q_\theta$ incluye la contribución de $\theta$, pero en la práctica, es necesario restar la anomalía verdadera $f$ para obtener el argumento del periapsis.

En mecánica celeste, el argumento del periapsis $w$ se define como el desplazamiento angular entre la dirección del nodo ascendente y el periapsis, medido en el plano orbital.

Como la posición angular del asteroide en su órbita en un instante dado es $\theta$, y que esta incluye tanto el argumento del periapsis $\omega$ como la anomalía verdadera $f$, se tiene:

$$
Q_\theta = \theta - f = \omega
$$

$$
\boxed{
  Q_\theta = \omega
}
$$

**$→$** Por último, se halla el Hamiltoniano del sistema.

La ecuación de Hamilton-Jacobi se expresa como:

$$
H \left(q_i, \frac{\partial S}{\partial q_i} \right) + \frac{\partial S}{\partial t} = 0
$$

Analizando los términos individualmente, se tiene que:

$$
\frac{\partial S}{\partial t} = -P_r
$$

Dado que en los antecedentes del problema se definió que la energía del sistema es:

$$
H = E = P_r
$$

Al sustituir esta relación en la ecuación de Hamilton-Jacobi, obtenemos:

$$
P_r - P_r = 0
$$

Esto implica que la función generatriz (la acción) satisface:

$$
H \left(q_i, \frac{\partial S}{\partial q_i} \right) + \frac{\partial S}{\partial t} = 0
$$

Por lo tanto, podemos redefinir el Hamiltoniano en términos de las nuevas variables canónicas $Q_r, Q_\theta, P_r, P_\theta$ como:

$$
\boxed{
H' (Q_r, Q_\theta, P_r, P_\theta) = 0
}
$$

Este resultado muestra que el Hamiltoniano transformado $H'$ es nulo en el nuevo sistema de coordenadas, lo que es característico de una transformación canónica donde una de las nuevas coordenadas actúa como un tiempo generalizado.

De esta manera, se ha demostrado:

$$
\boxed{
\begin{aligned}
  Q_r  & = -t_p\\
  Q_\theta & = \omega\\
  P_r & = -\frac{m_r\mu}{2a}\\
  P_\theta & = m_r\sqrt{\mu a(1-e^2)}\\
  H'(Q_r,Q_\theta,P_r,P_\theta) & = 0
  \end{aligned}
}
$$


### **Demostración 4.**

En este punto se demostrará cómo la transformación canónica con la función generatriz de tipo 1 nos lleva a las variables de Delaunay y a un nuevo Hamiltoniano.

Primero, la función generatriz dada es:

$$
F_{qP}(Q_r, Q_\theta, L_D, G_D) = \left( nL_D - \frac{3\mu m_r}{2a} \right) (t + Q_r) + Q_\theta G_D = F
$$

donde $n$ es la frecuencia media orbital, definida como:

$$
n = \sqrt{\frac{\mu}{a^3}}
$$

y $a$ es el semieje mayor de la órbita.

Dado que esta es una función generatriz de tipo 1 $F_1$, tenemos la relación:

$$
p_i = \frac{\partial F_1}{\partial q_i}, \quad l_i = \frac{\partial F_1}{\partial P_i}
$$

donde:

- $p_i = (P_r, P_\theta)$ son los momentos en la base original.
- $q_i = (Q_r, Q_\theta)$ son las coordenadas generalizadas originales.
- $l_i = (l_D, g_D)$ son las nuevas coordenadas canónicas.
- $P_i = (L_D, G_D)$ son los nuevos momentos canónicos conjugados.

**Las nuevas coordenadas se obtienen como:**
\begin{equation}
l_D = \frac{\partial F}{\partial L_D}, \quad g_D = \frac{\partial F}{\partial G_D}
\end{equation}

Se calcula cada derivada:
\begin{equation}
l_D = \frac{\partial}{\partial L_D} \left[ \left( nL_D - \frac{3\mu m_r}{2a} \right)(t + Q_r) + Q_\theta G_D \right]
\end{equation}

Como $nL_D - \frac{3\mu m_r}{2a}$ multiplica a $(t+Q_r)$, su derivada con respecto a $L_D$ es simplemente $n(t+Q_r)$:
\begin{equation}
l_D = n(t+Q_r)
\end{equation}

Por la definición de la anomalía media $M = n (t - t_p)$, identificamos que:

$$
\boxed{
\begin{aligned}
  l_D = M
\end{aligned}
}
$$

Ahora, se deriva con respecto a $G_D$:
\begin{equation}
g_D = \frac{\partial}{\partial G_D} \left[ \left( nL_D - \frac{3\mu m_r}{2a} \right)(t + Q_r) + Q_\theta G_D \right]
\end{equation}

Como el único término que depende de $G_D$ es $Q_\theta G_D$, obtenemos:

$$
g_D = Q_\theta
$$

Se sabe que $Q_θ$ representa el argumento del periastro $\omega$, así que:

$$
\boxed{
\begin{equation}
g_D = \omega
\end{equation}
}
$$

Ahora, se calcula **los nuevos momentos canónicos:**
\begin{equation}
P_{r} = L_D = \frac{\partial F}{\partial Q_r}, \quad P_{\theta} = G_D = \frac{\partial F}{\partial Q_\theta}
\end{equation}

Luego, se deriva con respecto a $Q_r$:
\begin{equation}
L_D = \frac{\partial}{\partial Q_r} \left[ \left( nL_D - \frac{3\mu m_r}{2a} \right)(t + Q_r) + Q_\theta G_D \right]
\end{equation}

Dado que el primer término es lineal en $Q_r$, su derivada es simplemente el coeficiente:
\begin{equation}
L_D = nL_D - \frac{3\mu m_r}{2a}
\end{equation}

Se usan la siguiente definición:
\begin{equation}
L_D = P_r = -\frac{m_r \mu}{2a}
\end{equation}

Se reemplaza y se tiene que:

$$ n L_D - \frac{m_r 3 \mu}{2a} = -\frac{m_r \mu}{2a} $$

$$ n L_D = \frac{2 m_r \mu}{2a} = \frac{m_r \mu}{a} $$

Y con la definición de frecuencia orbital media $n$, se tiene:

$$ n^{-1} = \sqrt{\frac{a^3}{\mu}} $$

Así:

$$ L_D = \frac{m_r \mu}{a} \sqrt{\frac{a^3}{\mu}} = \frac{m_r a}{a} \sqrt{\frac{a^2 \mu^2}{\mu}}$$

Y se obtiene que:

$$
\boxed{
\begin{equation}
L_D = m_r \sqrt{\mu a}
\end{equation}
}
$$

Ahora, para $P_{\theta}$:

\begin{equation}
P_\theta = G_D = \frac{\partial}{\partial Q_\theta} \left[ \left( nL_D - \frac{3\mu m_r}{2a} \right)(t + Q_r) + Q_\theta G_D \right]
\end{equation}

Como solo el término $Q_\theta G_D$ depende de $Q_\theta$, su derivada es simplemente:
\begin{equation}
P_{\theta} = G_D
\end{equation}

Dado que $P_\theta = m_r h$ y $h = \sqrt{\mu a (1 - e^2)}$, la expresión para el momento canónico $G_D$ es:

$$
\boxed{
\begin{equation}
G_D = m_r \sqrt{\mu a (1 - e^2)}
\end{equation}
}
$$

Ahora, se verifica el Hamiltoniano, para esto, se toma el Hamiltoniano del punto anterior, el cual es $H' = 0$, y se tiene la siguiente formula para hallar el nuevo $K$ así:

$$ K = H' + \frac{\partial F}{\partial t} $$

$$ K = \frac{\partial F}{\partial t} $$

Derivando con respecto al tiempo $t$:

$$
\frac{\partial F_{qP}}{\partial t} = nL_D - \frac{3\mu m_r}{2a}
$$

Usando que $nL_D = \frac{\mu m_r}{a}$, tenemos:

$$
K = \frac{\mu m_r}{a} - \frac{3\mu m_r}{2a} ⇒ K = \frac{2\mu m_r}{2a} - \frac{3\mu m_r}{2a}
$$


$$
\boxed{
\begin{equation}
K = -\frac{\mu m_r}{2a}
\end{equation}
}
$$

Puesto que la transformación a las variables de Delaunay es canónica, el nuevo Hamiltoniano $K$ conserva la misma estructura.

De esta manera, se ha demostrado que:

  $$
  \boxed{
  \begin{aligned}
  l_D  & = M\\
  g_D & = \omega\\
  L_D & = m_r\sqrt{\mu a}\\
  G_D & = m_r\sqrt{\mu a(1-e^2)}\\
  K(l_D,g_D,L_D,G_D) & = -\frac{\mu m_r}{2a}
  \end{aligned}
  }
  $$


### **Demostración 5.**

El Hamiltoniano dado es:

$$
H = -\frac{1}{4} \frac{GM m_r}{a} (1 - e^2)^{-3/2} \sum_{i=1}^{4} \frac{m_i a_i^2}{M a^2}
$$

donde:

- $m_r$ es la masa reducida del asteroide
- $a$ es el semieje mayor de la órbita del asteroide
- $e$ es la excentricidad
- $M$ es la masa del Sol
- $m_i$ y $a_i$ son las masas y semiejes mayores de los planetas perturbadores

Las variables de Delaunay están definidas como:

$$
L_D = m_r \sqrt{\mu a}, \quad G_D = m_r \sqrt{\mu a(1 - e^2)}, \quad g_D = \omega, \quad l_D = M
$$

Se despeja $a$ y $e$ en función de $L_D$ y $G_D$:

$$
a = \frac{L_D^2}{\mu m_r^2}, \quad (1 - e^2) = \frac{G_D^2}{L_D^2}
$$

Luego, se reemplaza $a$ y $e$ en el Hamiltoniano:

$$
H = -\frac{1}{4} \frac{GM m_r}{\left( \frac{L_D^2}{\mu m_r^2} \right)} \left( \frac{G_D^2}{L_D^2} \right)^{-3/2} \sum_{i=1}^{4} \frac{m_i a_i^2}{M \left( \frac{L_D^2}{\mu m_r^2} \right)^2}
$$

Se reemplaza $\mu = GM$, se opera y simplifica:

$$
H = -\frac{1}{4} \frac{GM m_r^3 (GM)}{L_D^2} \left( \frac{L_D^3}{G_D^3} \right) \sum_{i=1}^{4} \frac{m_i a_i^2 (GM)^2 m_r^4}{M L_D^4}
$$

$$
H = -\frac{1}{4} \frac{G^2 M^2 m_r^3 L_D}{G_D^3} \cdot \frac{G^2 M^2 m_r^4}{M L_D^4} \sum_{i=1}^{4} m_i a_i^2
$$

Así, tenemos el Hamiltoniano del asteroide en términos de las variables de Delaunay $L_D, G_D$ es:

$$
\boxed{
H = -\frac{1}{4} \frac{G^4 M^3 m_r^7}{G_D^3 L_D^3} \sum_{i=1}^{4} m_i a_i^2
}
$$

Ahora, se plantean las ecuaciones canónicas así:

- $\dot{l}_D= \frac{\partial H}{\partial L_D}$

- $\dot{g}_D = \frac{\partial H}{\partial G_D} $

- $\dot{L}_D= \frac{\partial H}{\partial l_D}$

- $\dot{G}_D = \frac{\partial H}{\partial g_D} $

Luego, para hallar la velocidad de avance del periapsis $\dot w$ se usa la ecuación $g_D = w$:

$$ \dot g_D = \frac{\partial H}{\partial G_D} = \dot w $$

Entonces,

$$ \frac{\partial H}{\partial G_D} =  \frac{\partial}{\partial G_D}\left(-\frac{1}{4}\frac{G^4M^3m_r^7}{L_D^3G_D^3}\sum_{i=1}^4 m_i a_i^2 \right)$$

$$ \frac{\partial H}{\partial G_D} = \frac{3}{4}\frac{G^4M^3m_r^7}{L_D^3G_D^4}\sum_{i=1}^4 m_i a_i^2 = \dot{w}$$

$$
\boxed{
  \dot{w} = \frac{3}{4}\frac{G^4M^3m_r^7}{L_D^3G_D^4}\sum_{i=1}^4 m_i a_i^2
}
$$

Ahora, el periodo de precesión $P_w$ está dado por:

$$ P_w = \frac{2 \pi}{\dot w} $$

Reemplazando $\dot w$ se tiene:

$$P_w =2 \pi \frac{4}{3}\frac{L_D^3G_D^4}{G^4M^3m_r^7}\sum_{i=1}^4 \frac{1}{m_i a_i^2}$$

El período orbital en términos de los elementos orbitales es:
\begin{equation}
    P = 2\pi \sqrt{\frac{a^3}{\mu}}
\end{equation}

Así, la relación entre los períodos se expresa como:
\begin{equation}
    \frac{P_\omega}{P} = 2\pi \frac{4}{3} \frac{L_D^3 G_D^4}{G^4 M^3 m_r^7} \cdot \frac{\sqrt{\mu}}{2\pi \sqrt{a^3}} \sum_{i=1}^{4} \frac{1}{m_i a_i^2}
\end{equation}

Simplificando se tiene:
\begin{equation}
    \frac{P_\omega}{P} = \frac{4}{3} \frac{L_D^3 G_D^4}{G^4 M^3 m_r^7} \sqrt{\frac{\mu}{a^3}} \sum_{i=1}^{4} \frac{1}{m_i a_i^2}
\end{equation}

Se expresa $\sqrt{\mu / a^3}$ en términos de $L_D$ y $m_r$:
\begin{equation}
    \sqrt{\frac{\mu}{a^3}} = \frac{\mu^{1/2}}{\left( \frac{L_D^2}{m_r^2 \mu} \right)^{3/2}}
\end{equation}

Sustituyendo:
\begin{equation}
    \frac{P_\omega}{P} = \frac{4}{3} \frac{L_D^3 G_D^4}{G^4 M^3 m_r^7} \cdot \frac{\mu^2 m_r^3}{L_D^3} \sum_{i=1}^{4} \frac{1}{m_i a_i^2}
\end{equation}

Cancelando $L_D^3$ se obtiene que:
\begin{equation}
    \frac{P_\omega}{P} = \frac{4}{3} \frac{G_D^4}{\mu^2 m_r^4} \sum_{i=1}^{4} \frac{M}{m_i a_i^2}
\end{equation}

Se usa la relación $G_D^2 = L_D^2 (1 - e^2)$:
\begin{equation}
    \frac{P_\omega}{P} = \frac{4}{3} \frac{G_D^4}{L_D^4} \sum_{i=1}^{4} \frac{M a^2}{m_i a_i^2}
\end{equation}

Finalmente, con $G_D^4 / L_D^4 = (1 - e^2)^2$:

$$
\boxed{
\begin{equation}
    \frac{P_\omega}{P} = \frac{4}{3} (1 - e^2)^2 \sum_{i=1}^{4} \frac{M a^2}{m_i a_i^2}
\end{equation}
}
$$
